```python
df_cancel['Appointment Start Date']=pd.to_datetime(df_cancel['Appointment Start Date'])
df_cancel['Appointment Cancel Date']=pd.to_datetime(df_cancel['Appointment Cancel Date'])
```

```python
df_cancel['Cancellation Lead'] = df_cancel['Appointment Start Date'] - df_cancel['Appointment Cancel Date']
```

```python
df_cancel['Cancellation Lead']
```

```python
df_cancel['Cancellation Lead'].sort_values()
```

```python
for df_cancel['Cancellation Lead']<=1, df_cancel['Cancellation Lead Series'] = 1
for df_cancel['Cancellation Lead'] > 1 and df_cancel['Cancellation Lead'] <=3, df_cancel['Cancellation Lead Series'] = 3
for df_cancel['Cancellation Lead'] > 3 and df_cancel['Cancellation Lead'] <=5, df_cancel['Cancellation Lead Series'] = 5
for df_cancel['Cancellation Lead'] > 5, df_cancel['Cancellation Lead Series'] = 7
```

```python
from datetime import date, timedelta
```

```python
if df_cancel['Cancellation Lead'] <= timedelta(days=1):
    df_cancel['Cancellation Lead Series'] = 1
```

```python
type(df_cancel['Cancellation Lead'])
```

```python
df_cancel['Cancellation Lead String'] = df_cancel['Cancellation Lead'].astype(str)
```

```python
df_cancel['Cancellation Lead Series'] = df_cancel['Cancellation Lead String'].str[0]
```

```python
dates = ['2023-08-14', '2023-08-29', '2023-09-13', '2023-09-22', '2023-10-02', '2023-10-27', '2023-11-01', 
              '2023-12-05']
#Subdivide into <1 day, 2-3, 4-5, >6
group1 = [5, 0, 0, 0, 14, 0, 1, 10]
group23 = [0, 0, 3, 0, 15, 3, 0, 0]
group45 = [0, 3, 0, 2, 1, 0, 0, 0]
group6 = [0, 0, 0, 0, 28, 0, 0, 0]
```

```python
x = np.arange(len(dates))
plt.bar(x, group1, label='One Day Notice', color='skyblue')
plt.bar(x, group23, bottom=group1, label='Two or Three Day Notice', color='orange')
plt.bar(x, group45, bottom=np.array(group1) + np.array(group23), label='Four or Five Day Notice', color='green')
plt.bar(x, group6, bottom=np.array(group1) + np.array(group23) + np.array(group45), label='6 or More Day Notice', color='black')

plt.title('Days Ahead of Time Appointments Were Cancelled on Each Day')
plt.xticks(x, dates, rotation=45)
plt.legend()

plt.show()
```
